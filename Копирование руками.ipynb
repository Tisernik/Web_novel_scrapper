```python
import pyperclip
import time
import pyautogui
from pathlib import Path
import os
import re
import random
```

```python
'''
1. Ожидание настройки сайта (таймаут 20 секунд)
2. Копирование содержимого сайта
3. Очистка текста от посторонних символов
3. Сохранение из буфера обмена текста в файл
4. Окончание сохранения глав => разметка текста
5. Приведение данных в модель для fb2
6. Создание fb2 файла

'''
```

```python
# Переменные

# Количество глав на странице
chapter_cnt = 1486
```

```python


def clean_text(text):
    # Удаляем все символы, которые не являются печатаемыми символами ASCII
    return ''.join(char for char in text if char.isascii())


time.sleep(5)  # Задержка на 5 секунд


# Получаем текущий рабочий каталог
current_dir = Path.cwd()  # Или Path('.').resolve(), если требуется абсолютный путь

# Формируем путь к папке "files"
files_folder = current_dir / 'Swallowed Star'

# Проверяем существование папки "files". Если её нет, создаём
if not files_folder.exists():
    files_folder.mkdir(parents=True)

# Создаём файлы в цикле
range_chapter = chapter_cnt
for i in range(chapter_cnt):
    filename = f"Глава_{range_chapter}.txt"

    pyautogui.hotkey('ctrl', 'a')
    pyautogui.hotkey('ctrl', 'c')
    
    # Формируем полный путь к файлу
    file_path = files_folder / filename

    # Чтение данных из буфера обмена
    
    clipboard_content = pyperclip.paste()
    # Очистка текста от посторонних символов
    clean_text_ = clean_text(clipboard_content)
    
    with open(file_path, 'w') as file:
        file.write(clean_text_)
    time_to_sleep = random_number = random.randint(1, 2)*2
    
    time.sleep(time_to_sleep)  # Задержка на 5 секунд

    range_chapter = range_chapter - 1

    if range_chapter < range_chapter - 10:
        break

print("Содержимое буфера обмена успешно сохранено в файл 'buffer_contents.txt'.")
```
