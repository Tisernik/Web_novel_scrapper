```python
import pyperclip
import time
import pyautogui
from pathlib import Path
import os
import re
import random
from glob import glob
import sys
from IPython.display import clear_output


def clean_text(text):
    # Удаляем все символы, которые не являются печатаемыми символами ASCII
    return ''.join(char for char in text if char.isascii())

def update_progress(progress):
    bar_length = 20
    if isinstance(progress, int):
        progress = float(progress)
    if not isinstance(progress, float):
        progress = 0
    if progress < 0:
        progress = 0
    if progress >= 1:
        progress = 1

    block = int(round(bar_length * progress))
    clear_output(wait = True)
    text = "Progress: [{0}] {1:.1f}%".format( "#" * block + "-" * (bar_length - block), progress * 100)
    print(text)


def cleaning_text(input_text):
    content = input_text.replace('  ', ' ')
    content = re.sub(r'\n\n\n', r'\n', content)
    content = re.sub(r'\n\n', r'\n', content)
    content = content.split('Next Chapter')
    content = content[1]
    content = content.split('Prev Chapter')
    content = content[0]
    if content.startswith('\n'):
        content = content[1:]

    if content.endswith('\n'):
        content = content[:-1]
    
    return content

```

```python
last_chapter_num = 1429 # Номер последней главы для парсинга
first_chapter_num = 1 # Номер первой главы для парсинга
book_name = 'Swallowed Star'

range_chapter = last_chapter_num - first_chapter_num + 1
```

```python
# Получаем текущий рабочий каталог
current_dir = Path.cwd()  # Или Path('.').resolve(), если требуется абсолютный путь

# Формируем путь к папке "Swallowed Star"
files_folder = current_dir / book_name

# Проверяем существование папки "files". Если её нет, создаём
if not files_folder.exists():
    files_folder.mkdir(parents=True)
```

```python
time.sleep(5)  # Задержка на 5 секунд чтобы успеть переместиться на активное окно

current_chapter = last_chapter_num
for i in range(range_chapter):
    filename = f"Глава_{current_chapter}.txt"

    pyautogui.hotkey('ctrl', 'a')
    pyautogui.hotkey('ctrl', 'c')
    
    # Формируем полный путь к файлу
    file_path = files_folder / filename

    # Чтение данных из буфера обмена
    clipboard_content = pyperclip.paste()
    
    # Очистка текста от посторонних символов
    clean_text_ = clean_text(clipboard_content)
    
    with open(file_path, 'w') as file:
        file.write(clean_text_)
    
    time_to_sleep = random_number = random.randint(1, 2)*2
    time.sleep(time_to_sleep)  # Задержка на 5 секунд

    current_chapter -= 1

    if range_chapter < range_chapter - 10:
        break

print("Содержимое буфера обмена успешно сохранено в файл")
```
